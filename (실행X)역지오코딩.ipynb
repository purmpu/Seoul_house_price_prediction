```python
import pandas as pd
import time
from geopy.geocoders import Nominatim

# 파일 불러오기
df = pd.read_csv("seoul - SeoulRealEstate.csv")

# 지오코더 초기화
geolocoder = Nominatim(user_agent='South Korea', timeout=None)

# 역지오코딩 함수 정의
def geocoding_reverse(lat, lng):
    try:
        location = geolocoder.reverse(f"{lat}, {lng}")
        if location and 'address' in location.raw:
            return location.address
        else:
            return None
    except Exception as e:
        return None

# 주소 열 추가 (1초 딜레이 포함)
addresses = []
for i, row in df.iterrows():
    lat, lng = row['lat'], row['lng']
    addr = geocoding_reverse(lat, lng)
    addresses.append(addr)
    print(f"{i+1}/{len(df)} : {addr}")
    time.sleep(1)  # 서버 과부하 방지

df['address'] = addresses

# 저장
df.to_csv("seoul_with_address.csv", index=False, encoding='utf-8-sig')
```
